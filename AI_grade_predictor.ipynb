Împărțirea datelor în test și antrenament

```python
import pandas as pd

data_frame = pd.read_csv('student_habits_performance.csv')
#print(data_frame.head())
print(data_frame.info())

from sklearn.model_selection import train_test_split

y = data_frame['exam_score']

X = data_frame.drop(columns = ['exam_score'], axis = 1)

#împart datele 80% in setul de antrenament si 20% in setul de testare
X_train, X_test, y_train, y_test = train_test_split(X, data_frame['exam_score'],
                                   test_size = 0.2, random_state=42)
print(f"Dimensiunea setului de antrenament: {X_train.shape}")
print(f"Dimensiunea setului de test: {X_test.shape}")

# unesc datele din X si y si salvez rezultatele in format csv
train_data_frame = X_train.join(y_train);
test_data_frame = X_test.join(y_test);
train_data_frame.to_csv('train.csv', index=False)
test_data_frame.to_csv('test.csv', index=False)
```

Procesarea seturilor de date pentru a putea fi interpretate de modelul de învățare

```python
# elimin coloanele cu educatia parintilor si cu id-ul
X_train = X_train.drop(
          columns = ['student_id','parental_education_level'], axis=1)
X_test = X_test.drop(
         columns = ['student_id','parental_education_level'], axis=1)

from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()

# aplic Label Encoding pe coloanele gender, part_time_job,
# extracurricular_participation

X_train['gender'] = encoder.fit_transform(X_train['gender'])
X_test['gender'] = encoder.fit_transform(X_test['gender'])
# Female = 0, Male = 1

X_train['part_time_job'] = encoder.fit_transform(X_train['part_time_job'])
X_test['part_time_job'] = encoder.fit_transform(X_test['part_time_job'])
# yes = 1, no = 0

X_test['extracurricular_participation'] = encoder.fit_transform(
                                        X_test['extracurricular_participation'])
X_train['extracurricular_participation'] = encoder.fit_transform(
                                      X_train['extracurricular_participation'])
# yes = 1, no = 0


#aplic One-Hot Encoding pe coloanele diet_quality și internet_quality

X_train = pd.get_dummies(X_train, columns = ['diet_quality'], drop_first=True)
X_test = pd.get_dummies(X_test, columns = ['diet_quality'], drop_first=True)

X_train = pd.get_dummies(
          X_train, columns = ['internet_quality'], drop_first=True)
X_test = pd.get_dummies(X_test, columns = ['internet_quality'], drop_first=True)



from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# salvez originalul pentru data_frame-ul de test inainte de standardizare
X_input_test = X_test

# standardizări
X_train[['age', 'study_hours_per_day', 'social_media_hours',
        'netflix_hours', 'attendance_percentage', 'sleep_hours']] =(
        scaler.fit_transform(
        X_train[['age','study_hours_per_day', 'social_media_hours',
                 'netflix_hours', 'attendance_percentage', 'sleep_hours']]))

X_test[['age', 'study_hours_per_day', 'social_media_hours',
        'netflix_hours', 'attendance_percentage', 'sleep_hours']] = (
        scaler.fit_transform(
        X_test[['age','study_hours_per_day', 'social_media_hours',
                'netflix_hours', 'attendance_percentage', 'sleep_hours']]))


print(X_train.head(20));
```

Antrenarea modelului și evaluarea acestuia

```python
#antrenarea modelului
from sklearn.linear_model import LinearRegression

# Creăm modelul
model = LinearRegression()

# Antrenăm modelul pe setul de antrenament
model.fit(X_train, y_train)

# Prezicem notele pentru datele de test
y_pred = model.predict(X_test)

plt.scatter(y_test,y_pred)

import numpy as np
# elimin valorile mai mari decât 100
y_pred = np.clip(y_pred, 0, 100)
plt.scatter(y_test,y_pred)

from sklearn.metrics import root_mean_squared_error

# Calculăm eroarea pătratică medie
rmse = root_mean_squared_error(y_test, y_pred)
print(f"root_mean Squared Error: {rmse:.2f}")
```

Procesare input utilizator cu Gradio și funcția specială

```python
def predict_grade(age, study_hours_per_day, social_media_hours,
                  netflix_hours, attendance_percentage, sleep_hours,
                  mental_health_rating):
    global X_input_test
    input_utilizator = np.array([[age, 1, study_hours_per_day,
                       social_media_hours, netflix_hours, 0,
                       attendance_percentage, sleep_hours, 3,
                      mental_health_rating, 0, 1, 0, 1, 0]])
    # transform vectorul intr-un data frame cu o singură linie
    input_utilizator = pd.DataFrame(input_utilizator,
                                    columns=X_input_test.columns)
    # concatenez linia la datele de test
    X_input_test = pd.concat([X_input_test, input_utilizator],
                             ignore_index=True)

    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    # standardizez datele de test, inclusiv linia concatenată
    X_input_test[['age', 'study_hours_per_day', 'social_media_hours',
                  'netflix_hours', 'attendance_percentage', 'sleep_hours']] =(
                  scaler.fit_transform(
                  X_input_test[['age','study_hours_per_day',
                  'social_media_hours', 'netflix_hours',
                  'attendance_percentage', 'sleep_hours']]))
    # extrag linia standardizată de la final
    input_utilizator = X_input_test.tail(1)

    # fac predicția
    predictie = model.predict(input_utilizator)[0]
    predictie = np.clip(predictie, 0, 100)  # limitează între 0 și 100
    return f"Nota prezisă: {predictie:.2f}"

# creez interfața Gradio
interface = gr.Interface(
    fn=predict_grade,
    inputs=[
        gr.Number(label="vârstă"),
        gr.Number(label="ore de învățat pe zi"),
        gr.Number(label="ore petrecute pe social media"),
        gr.Number(label="ore pe netflix"),
        gr.Number(label="procent de prezențe aproximativ:"),
        gr.Number(label="ore de somn pe noapte"),
        gr.Number(label="starea ta de sănătate mentală(de la 1 la 10)")],
    outputs="text",
    title="Predicția notei elevului",
    description=("Introdu valorile pentru obiceiurile elevului și apasă Submit"
                 "pentru a vedea nota prezisă.")
)

interface.launch(debug=True)
```
